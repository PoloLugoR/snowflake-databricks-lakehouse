```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.gold")
spark.sql("CREATE VOLUME IF NOT EXISTS workspace.bronze.raw_files")
```

```python
import re

raw_path = "/Volumes/workspace/bronze/raw_files/DataCoSupplyChainDataset.csv"

def clean(name):
    return re.sub(r"[^0-9a-zA-Z]+", "_", name.strip().lower()).strip("_")

df = (
    spark.read
    .option("header", True)
    .option("encoding", "ISO-8859-1")
    .option("quote", '"')
    .option("escape", '"')
    .csv(raw_path)
)
df = df.toDF(*[clean(c) for c in df.columns])

(df.write
   .format("delta")
   .mode("overwrite")
   .saveAsTable("workspace.bronze.supply_chain_orders"))

print(spark.table("workspace.bronze.supply_chain_orders").count())
```
